```python
class Chocolate:
    def __init__(self, price: float, addition: str = ""):
        if price <= 0:
            raise ValueError("Цена шоколадного батончика должна быть больше нуля")
        self.price = float(price)
        self.addition = addition

    def show_my_chocolate(self):
        if self.addition:
            print(f"Шоколадный батончик, добавка: {self.addition}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return f"Шоколадный батончик, {self.price:g} рублей"
```

```python
chocolate_1 = Chocolate(100.0)
chocolate_2 = Chocolate(150.5, "орехи")
chocolate_3 = Chocolate(85.0, "карамель")

for chocolate in (chocolate_1, chocolate_2, chocolate_3):
    chocolate.show_my_chocolate()
    print(chocolate)
    print()
```

```python
for wrong_price in (-10, 0):
    try:
        Chocolate(wrong_price)
    except ValueError as error:
        print(f"Цена {wrong_price}: {error}")
```
