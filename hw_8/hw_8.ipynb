## 1. Подключение библиотек



```python
import os
from getpass import getpass

import pandas as pd
import requests
from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 80)
print("Библиотеки для работы с API загружены.")
```

## 2. API — получить данные обо всех книгах



```python
BASE_URL = "https://www.anapioficeandfire.com/api"
PAGE_SIZE = 50

session = requests.Session()
session.headers.update({"User-Agent": "Python-Data-Analysis-Homework/1.0", "Accept": "application/json"})

def load_all_pages(resource: str, extra_params=None):
    """Собрать все записи из постраничного API (GET-запросы)."""
    items = []
    for page in range(1, 101):  # защита от бесконечного цикла
        params = {"page": page, "pageSize": PAGE_SIZE}
        if extra_params:
            params.update(extra_params)
        response = session.get(f"{BASE_URL}/{resource}", params=params, timeout=30)
        response.raise_for_status()
        rows = response.json()
        if not isinstance(rows, list):
            raise TypeError(f"Неожиданный формат ответа {resource}, страница {page}: {type(rows)}")
        if not rows:
            break
        items.extend(rows)
        print(f"{resource}: страница {page}, получено {len(rows)} записей")
        if len(rows) < PAGE_SIZE:
            break
    else:
        raise RuntimeError("Достигнут лимит в 100 страниц: проверьте пагинацию.")
    if not items:
        raise RuntimeError(f"API не вернул данные из ресурса {resource}.")
    return items
```

```python
books = load_all_pages("books")
books_df = pd.DataFrame(books)
print(f"Всего книг получено: {len(books_df)}")
display(books_df.head(10))
```

## 3. API — получить информацию обо всех домах Вестероса



```python
houses = load_all_pages("houses")
houses_df = pd.DataFrame(houses)
print(f"Всего домов получено: {len(houses_df)}")
display(houses_df.head(10))
```

## 4. API — получить дома, у которых есть титулы


```python
houses_with_titles = load_all_pages("houses", {"hasTitles": "true"})
houses_titles_df = pd.DataFrame(houses_with_titles)
print(f"Всего домов с титулами: {len(houses_titles_df)}")
if "titles" in houses_titles_df.columns:
    nonempty_titles = houses_titles_df["titles"].apply(lambda x: isinstance(x, list) and len(x) > 0)
    print("Дома с непустым полем titles:", int(nonempty_titles.sum()), "из", len(houses_titles_df))
display(houses_titles_df[[c for c in ["name", "region", "titles", "url"] if c in houses_titles_df.columns]].head(10))
```

## 5. Работа с PostgreSQL — подключение к публичной базе RNAcentral

Сведения о базе опубликованы на [официальной странице RNAcentral](https://rnacentral.org/help/public-database):

- Host: `hh-pgsql-public.ebi.ac.uk`
- Port: `5432`
- Database: `pfmegrnargs`
- User: `reader`

**Пароль не сохраняется в ноутбуке.** При запуске он запрашивается скрытым вводом, либо читается из переменной окружения `RNACENTRAL_PASSWORD`. 

```python
import psycopg2

DB_CONFIG = {
    "host": "hh-pgsql-public.ebi.ac.uk",
    "port": 5432,
    "dbname": "pfmegrnargs",
    "user": "reader",
    "connect_timeout": 15,
}
password = os.getenv("RNACENTRAL_PASSWORD") or getpass("Введите публичный пароль RNAcentral с официальной страницы: ")

conn = psycopg2.connect(**DB_CONFIG, password=password)
print("Подключение к RNAcentral PostgreSQL успешно установлено.")
```

## 6. Получить 10 строк из таблицы `rnc_database`



```python
with conn.cursor() as cursor:
    cursor.execute("SELECT * FROM rnc_database LIMIT 10")
    rows_all = cursor.fetchall()
    columns_all = [column[0] for column in cursor.description]

rnc_df = pd.DataFrame(rows_all, columns=columns_all)
print("Получено записей:", len(rnc_df))
display(rnc_df)
```

## 7. Получить четыре конкретных столбца для 10 записей


```python
query = """
SELECT display_name, num_sequences, num_organisms, url
FROM rnc_database
LIMIT 10
"""
with conn.cursor() as cursor:
    cursor.execute(query)
    selected_rows = cursor.fetchall()
    selected_cols = [column[0] for column in cursor.description]

selected_df = pd.DataFrame(selected_rows, columns=selected_cols)
print("Получено записей:", len(selected_df))
display(selected_df)

conn.close()
print("Соединение с PostgreSQL закрыто.")
```
