## 4. Получение прогноза стоимости


````python
import joblib

loaded_model = joblib.load("model.joblib")
new_flat = pd.DataFrame([{
    "city": "Москва",
    "total_square": 55.0,
    "rooms": 2,
    "floor": 7,
}])
predicted_price = float(loaded_model.predict(new_flat)[0])
print(f"Прогноз для квартиры площадью 55 м²: {predicted_price:,.0f} ₽".replace(",", " "))
````
